```python
%matplotlib widget
import matplotlib.image as mpimg
import matplotlib.pyplot as plt
from matplotlib.widgets import Button
line_groups = []
anti_line_groups = []
order = []

newLineGroup = True
newAntiLineGroup = False
mode = 0 # 0 - line, 1 - anti-line

def buttonClickPrep():
  global line_groups
  global anti_line_groups
  if mode == 1:
    if len(anti_line_groups) > 0:
      line = anti_line_groups[-1]
      line.pop()
      anti_line_groups[-1] = line
  else:
    if len(line_groups) > 0:
      line = line_groups[-1]
      line.pop()
      line_groups[-1] = line

def reDraw():
  global line_groups
  global anti_line_groups
  global imgplot
  global ax
  global fig
  if not (len(line_groups) > 0 and line_groups[-1] == []) or not (len(anti_line_groups) > 0 and anti_line_groups[-1] == []):
    ax.clear()
    img=mpimg.imread('/content/drive/MyDrive/BE 3090/GreenBirdYellowBeeYellowTree.JPG')
    imgplot = ax.imshow(img)
    for i in line_groups:
        ax.plot(list(map(lambda x: x[0],i)),list(map(lambda x: x[1],i)),color='green',marker='o')
    for i in anti_line_groups:
        ax.plot(list(map(lambda x: x[0],i)),list(map(lambda x: x[1],i)),color='red',marker='x')
    fig.canvas.draw_idle() # essential
    fig.canvas.flush_events() # essential

def deletePoint(event):
  global line_groups
  global anti_line_groups
  global newLineGroup
  global newAntiLineGroup
  global mode
  if mode == 1:
    if len(anti_line_groups) > 0:
      line = anti_line_groups[-1]
      line.pop()
      if len(line) > 0:
        line.pop()
      anti_line_groups[-1] = line
    else:
      deleteLine(None)
  else:
    if len(line_groups) > 0:
      line = line_groups[-1]
      line.pop()
      if len(line) > 0:
        line.pop()
      line_groups[-1] = line
    else:
      deleteLine(None)

  reDraw()

def deleteLine(event):
  global line_groups
  global anti_line_groups
  global newLineGroup
  global newAntiLineGroup
  global mode
  global order
  buttonClickPrep()
  if not (order == []):
    if order[-1] == 1:
      if anti_line_groups[-1] == []:
        anti_line_groups.pop()
        order.pop()
      elif anti_line_groups != [] and order[-1] == 1 :
        anti_line_groups.pop()
        order.pop()
      elif anti_line_groups != [] and order[-1] == 0 :
        line_groups.pop()
        order.pop()
      if order == []:
        mode = 0
      else:
        mode = order[-1]
    else:
      if line_groups[-1] == []:
        line_groups.pop()
        order.pop()
      elif line_groups != [] and order[-1] == 0 :
        line_groups.pop()
        order.pop()
      elif line_groups != [] and order[-1] == 1 :
        anti_line_groups.pop()
        order.pop()
      if order == []:
        mode = 0
      else:
        mode = order[-1]
  else:
    newLineGroup = True
    newAntiLineGroup = False
    mode = 0
  if order == []:
    newLineGroup = True
    newAntiLineGroup = False
    mode = 0
  reDraw()

def newLine(event):
  global newLineGroup
  global newAntiLineGroup
  global mode
  newLineGroup = True
  newAntiLineGroup = False
  buttonClickPrep()
  reDraw()

def newAntiLine(event):
  global newLineGroup
  global newAntiLineGroup
  global mode
  global ax
  newAntiLineGroup = True
  newLineGroup = False
  buttonClickPrep()
  reDraw()

def onclick(event):
    global line_groups
    global anti_line_groups
    global newLineGroup
    global newAntiLineGroup
    global imgplot
    global ax
    global buttonClick
    global fig
    global mode
    ix, iy = event.xdata, event.ydata
    if newLineGroup:
      mode = 0
      order.append(mode)
      if len(line_groups) > 0 and line_groups[-1] == []:
        line_groups.pop()
      line_groups.append([(ix,iy)])
      newLineGroup = False
    elif newAntiLineGroup:
      mode = 1
      order.append(mode)
      if len(anti_line_groups) > 0 and anti_line_groups[-1] == []:
        anti_line_groups.pop()
      anti_line_groups.append([(ix,iy)])
      newAntiLineGroup = False
    elif mode == 0:
      line = line_groups[-1]
      line.append((ix,iy))
      line_groups[-1] = line
    else:
      line = anti_line_groups[-1]
      line.append((ix,iy))
      anti_line_groups[-1] = line
    for i in line_groups:
        ax.plot(list(map(lambda x: x[0],i)),list(map(lambda x: x[1],i)),color='green',marker='o')
    for i in anti_line_groups:
        ax.plot(list(map(lambda x: x[0],i)),list(map(lambda x: x[1],i)),color='red',marker='x')
    fig.canvas.draw_idle() # essential
    fig.canvas.flush_events() # essential

fig, ax = plt.subplots()
img=mpimg.imread('/content/drive/MyDrive/BE 3090/GreenBirdYellowBeeYellowTree.JPG')
imgplot = ax.imshow(img)
axesLine = plt.axes([0.81, 0.000001, 0.1, 0.075])
bnewLine = Button(axesLine, 'new line',color="green")
bnewLine.on_clicked(newLine)
axesAntiLine = plt.axes([0.61, 0.000001, 0.15, 0.075])
bnewAntiLine = Button(axesAntiLine, 'new anti line',color="red")
bnewAntiLine.on_clicked(newAntiLine)
axesDeletePoint = plt.axes([0.41, 0.000001, 0.15, 0.075])
bdeletePoint = Button(axesDeletePoint, 'delete point',color="grey")
bdeletePoint.on_clicked(deletePoint)
axesDeleteLine = plt.axes([0.21, 0.000001, 0.15, 0.075])
bdeleteLine = Button(axesDeleteLine, 'delete line',color="lightgrey")
bdeleteLine.on_clicked(deleteLine)
cid = fig.canvas.mpl_connect('button_press_event', onclick)
```
